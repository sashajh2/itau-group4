```python
import numpy as np
import matplotlib.pyplot as plt


def sum_conf_matrices(results, key='global', cm_key='best_cm_acc'):
    summed_cm = np.zeros((2, 2), dtype=int)
    for fold in results:
        cm = np.array(fold[key][cm_key])
        summed_cm += cm
    return summed_cm

def sum_stratified_conf_matrices(results, strat_key, cm_key='best_cm_acc'):
    summed_cm = np.zeros((2, 2), dtype=int)
    for fold in results:
        if strat_key in fold['stratified']:
            cm = np.array(fold['stratified'][strat_key][cm_key])
            summed_cm += cm
    return summed_cm

def plot_conf_matrix(cm, title):
    fig, ax = plt.subplots()
    cax = ax.matshow(cm, cmap='Blues')
    for (i, j), val in np.ndenumerate(cm):
        ax.text(j, i, f'{val}', ha='center', va='center', color='black', fontsize=14)
    ax.set_xlabel('Predicted', fontsize=12)
    ax.set_ylabel('Actual', fontsize=12)
    ax.set_xticks([0, 1])
    ax.set_yticks([0, 1])
    ax.set_title(title, fontsize=14)
    fig.colorbar(cax)
    plt.show()
```

```python
# GLOBAL: sum & display
global_cm_acc = sum_conf_matrices(results, key='global', cm_key='best_cm_acc')
global_cm_fpr = sum_conf_matrices(results, key='global', cm_key='best_cm_fpr')

plot_conf_matrix(global_cm_acc, 'Global Confusion Matrix (Best Accuracy Threshold)')
plot_conf_matrix(global_cm_fpr, 'Global Confusion Matrix (Best FPR Threshold)')
```

```python
# STRATIFIED
strats = [
        "gaussian_noise",
        "random_brightness",
        "camera_shake",
        "rolling_shutter",
        "chromatic_aberration",
        "poisson_noise",
        "gaussian_blur",
        "exposure_variation",
        "lens_distortion",
        "vignetting",
        "low_bitrate",
        "motion_blur",
        "salt_and_pepper",
        "color_quantization",
        "speckle_noise",
    ]

for condition in [True, False]:
    for strat in strats:
        strat_key = (strat, condition)
        strat_cm_acc = sum_stratified_conf_matrices(results, strat_key, cm_key='best_cm_acc')
        strat_cm_fpr = sum_stratified_conf_matrices(results, strat_key, cm_key='best_cm_fpr')

        plot_conf_matrix(strat_cm_acc, f'{strat}={condition} (Best Accuracy Threshold)')
        plot_conf_matrix(strat_cm_fpr, f'{strat}={condition} (Best FPR Threshold)')
```
